```python
import os
import pandas as pd
import matplotlib.pyplot as plt

def load_and_clean_data(file_path):
    """Loads dataset with error handling and basic data cleaning."""
    try:
        if not os.path.exists(file_path):
            raise FileNotFoundError(f"Dataset not found at {file_path}")
        
        df = pd.read_csv(file_path)
        print(f"[{file_path}] loaded successfully. Rows: {len(df)}, Columns: {len(df.columns)}")
        
        initial_count = len(df)
        df = df.dropna().drop_duplicates()
        print(f"Cleaned data: Removed {initial_count - len(df)} rows containing missing values or duplicates.")
        return df
    except Exception as e:
        print(f"Error loading or cleaning data: {e}")
        return None

def analyze_data(df):
    """Calculates key metrics and extracts meaningful insights."""
    print("\n--- DATA ANALYSIS METRICS ---")
    if df is not None and 'Sales' in df.columns:
        total_sales = df['Sales'].sum()
        avg_sales = df['Sales'].mean()
        print(f"Total Revenue / Sales: ${total_sales:,.2f}")
        print(f"Average Order Value: ${avg_sales:,.2f}")
    
    if df is not None and 'Category' in df.columns and 'Sales' in df.columns:
        top_category = df.groupby('Category')['Sales'].sum().idxmax()
        print(f"Top Performing Category: {top_category}")
    print("-----------------------------\n")

def create_visualizations(df):
    """Generates, displays, and saves professional charts."""
    if df is None:
        return
        
    os.makedirs('visualizations', exist_ok=True)
    plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')

    # Chart 1: Bar Chart - Sales by Product Category
    if 'Category' in df.columns and 'Sales' in df.columns:
        plt.figure(figsize=(8, 5))
        category_sales = df.groupby('Category')['Sales'].sum().sort_values(ascending=False)
        category_sales.plot(kind='bar', color='#2b5c8f', edgecolor='black', width=0.6)
        plt.title('Total Sales by Product Category', fontsize=14, fontweight='bold', pad=15)
        plt.xlabel('Product Category', fontsize=12)
        plt.ylabel('Total Sales ($)', fontsize=12)
        plt.xticks(rotation=30, ha='right')
        plt.grid(axis='y', linestyle='--', alpha=0.7)
        plt.tight_layout()
        
        path1 = os.path.join(os.getcwd(), 'visualizations', 'sales_by_category.png')
        plt.savefig(path1, dpi=300)
        plt.show()  # Jupyter screen par chart turant dikhayega
        plt.close()
        print(f"Saved: {path1}")

    # Chart 2: Line Chart - Monthly Sales Trend Over Time
    if 'Date' in df.columns and 'Sales' in df.columns:
        df_copy = df.copy()
        df_copy['Date'] = pd.to_datetime(df_copy['Date'], errors='coerce')
        df_copy = df_copy.dropna(subset=['Date'])
        monthly_sales = df_copy.resample('ME', on='Date')['Sales'].sum()
        
        plt.figure(figsize=(10, 5))
        plt.plot(monthly_sales.index.strftime('%Y-%m'), monthly_sales.values, 
                 marker='o', color='#e74c3c', linewidth=2.5, markersize=6)
        plt.title('Monthly Sales Trend Over Time', fontsize=14, fontweight='bold', pad=15)
        plt.xlabel('Month', fontsize=12)
        plt.ylabel('Total Sales ($)', fontsize=12)
        plt.xticks(rotation=45)
        plt.grid(True, linestyle='--', alpha=0.7)
        plt.tight_layout()
        
        path2 = os.path.join(os.getcwd(), 'visualizations', 'monthly_trend.png')
        plt.savefig(path2, dpi=300)
        plt.show()  # Jupyter screen par chart turant dikhayega
        plt.close()
        print(f"Saved: {path2}")

# --- Execution Pipeline ---
DATA_FILE = "sales_data.csv"
print("Step 1-2: Initializing Project & Pipeline...")
dataframe = load_and_clean_data(DATA_FILE)

if dataframe is not None:
    print("Step 3-4: Running Data Exploration & Analysis...")
    analyze_data(dataframe)
    
    print("Step 5: Generating Visualizations...")
    create_visualizations(dataframe)
    print("Pipeline Execution Completed Successfully!")
import pandas as pd
df = pd.read_csv("sales_data.csv")
print("Aapke CSV ke columns yeh hain:", df.columns.tolist())
```
